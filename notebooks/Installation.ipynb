# Installation
## Linux
SoundLight is written in Python, but it has to be run in a Linux or Mac distribution due to _NATTEN_ builds not being available for Windows systems: 
1. Install `ffmpeg`: `sudo apt install ffmpeg`.
1. Run `sudo pip install -r requirements.txt`.
1. Run `pip install --upgrade setuptools wheel`
1. Install NATTEN:
    - If running on a GPU, use `sudo pip install natten==0.17.1+torch240cu124 -f https://shi-labs.com/natten/wheels`, with `cu124` corresponding to your CUDA version. See [Shi-Labs.com](https://www.shi-labs.com/natten/) for more info.
    - If running on a CPU, use `pip install natten==0.17.1+torch240cpu -f https://shi-labs.com/natten/wheels`.
1. Install MADMOM with admin privileges: `sudo pip install --upgrade --no-deps --force-reinstall --quiet 'git+https://github.com/CPJKU/madmom.git'`
1. Install All-In-One: `sudo pip install allin1`

## Windows
If desired to run in Windows, complete these steps before installing as if in Linux:
1. Install `wsl`.

# Requirements
- Python 3.10
    - `sudo apt-get install python3.10 python3.10-dev`
- `sudo apt-get install build-essential`
